```python
from scipy import stats
import numpy as np

# Sample data
sample = [51, 53, 49, 50, 52, 48, 54, 50, 49, 51]
sample_mean = np.mean(sample)
sample_size = len(sample)

# Known population parameters
population_mean = 50
population_std = 2  # Known population standard deviation

# Calculate Z-score
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

# Calculate p-value
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))  # two-tailed test

# Print results
print(f"Sample Mean: {sample_mean}")
print(f"Z-score: {z_score:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The sample mean is significantly different from the population mean.")
else:
    print("Fail to reject the null hypothesis: No significant difference between the sample and population mean.")
```

```python
#Perform a two-tailed Z-test using Python and visualize the decision region on a plot
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Step 1: Sample data and population info
sample = [68, 70, 65, 72, 69, 71, 67, 70, 68, 69]
sample_mean = np.mean(sample)
sample_size = len(sample)

# Population parameters
population_mean = 68
population_std = 3  # Known population standard deviation

# Step 2: Z-test calculation
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))
p_value = 2 * (1 - norm.cdf(abs(z_score)))  # two-tailed

# Step 3: Hypothesis testing decision
alpha = 0.05
z_critical = norm.ppf(1 - alpha/2)  # critical z-value for two-tailed test

# Step 4: Display results
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Z-Score: {z_score:.4f}")
print(f"P-Value: {p_value:.4f}")
if abs(z_score) > z_critical:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")

# Step 5: Visualization
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, 0, 1)

plt.figure(figsize=(10, 5))
plt.plot(x, y, label='Standard Normal Distribution')

# Shade rejection regions
plt.fill_between(x, y, where=(x <= -z_critical), color='red', alpha=0.3, label='Rejection Region (α/2)')
plt.fill_between(x, y, where=(x >= z_critical), color='red', alpha=0.3)

# Plot z-score
plt.axvline(z_score, color='blue', linestyle='--', label=f'Z = {z_score:.2f}')
plt.axvline(-z_critical, color='black', linestyle='--', label=f'Z-critical = ±{z_critical:.2f}')
plt.axvline(z_critical, color='black', linestyle='--')

plt.title("Two-Tailed Z-Test Visualization")
plt.xlabel("Z-Score")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Create a Python function that calculates and visualizes Type 1 and Type 2 errors during hypothesis testing
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def visualize_type1_type2_errors(mu0, mu1, sigma, n, alpha=0.05):
    """
    mu0: mean under H0
    mu1: mean under H1
    sigma: population standard deviation
    n: sample size
    alpha: significance level (default = 0.05)
    """
    # Standard error
    se = sigma / np.sqrt(n)

    # Critical value for right-tailed test
    z_critical = norm.ppf(1 - alpha)
    critical_value = mu0 + z_critical * se

    # Calculate beta (Type II error)
    beta = norm.cdf((critical_value - mu1) / se)

    # X-axis range
    x = np.linspace(mu0 - 4*se, mu1 + 4*se, 1000)

    # PDFs
    y0 = norm.pdf(x, mu0, se)  # Null distribution
    y1 = norm.pdf(x, mu1, se)  # Alternative distribution

    # Plot distributions
    plt.figure(figsize=(10, 6))
    plt.plot(x, y0, label="H₀ Distribution", color='blue')
    plt.plot(x, y1, label="H₁ Distribution", color='green')

    # Shade Type I error region (α)
    x_alpha = x[x >= critical_value]
    plt.fill_between(x_alpha, norm.pdf(x_alpha, mu0, se), color='red', alpha=0.4, label='Type I Error (α)')

    # Shade Type II error region (β)
    x_beta = x[x < critical_value]
    plt.fill_between(x_beta, norm.pdf(x_beta, mu1, se), color='orange', alpha=0.4, label='Type II Error (β)')

    # Critical value line
    plt.axvline(critical_value, color='black', linestyle='--', label=f'Critical Value = {critical_value:.2f}')

    plt.title("Visualization of Type I and Type II Errors")
    plt.xlabel("Sample Mean")
    plt.ylabel("Probability Density")
    plt.legend()
    plt.grid(True)
    plt.show()

    # Print error probabilities
    print(f"Type I Error (α): {alpha}")
    print(f"Type II Error (β): {beta:.4f}")
    print(f"Power of the test (1 - β): {1 - beta:.4f}")

# Example usage
visualize_type1_type2_errors(mu0=100, mu1=105, sigma=15, n=30, alpha=0.05)
```

```python
#Write a Python function to calculate the confidence interval for a sample mean and explain its significance
import numpy as np
from scipy import stats

def confidence_interval(data, confidence=0.95):
    """
    Calculate the confidence interval for a sample mean.

    Parameters:
    - data: list or numpy array of sample values
    - confidence: confidence level (default = 0.95 for 95%)

    Returns:
    - Tuple: (lower bound, upper bound)
    """
    n = len(data)
    mean = np.mean(data)
    se = stats.sem(data)  # Standard Error of the Mean
    t_critical = stats.t.ppf((1 + confidence) / 2, df=n-1)  # Two-tailed t critical value

    margin_of_error = t_critical * se
    lower = mean - margin_of_error
    upper = mean + margin_of_error

    return (round(lower, 4), round(upper, 4))
# Sample data
sample_data = [22, 24, 20, 23, 21, 22, 25, 26, 24, 23]

# Calculate 95% Confidence Interval
ci = confidence_interval(sample_data, confidence=0.95)
print("95% Confidence Interval for the mean:", ci)
```

```python
#Write a Python program to calculate the margin of error for a given confidence level using sample data
import numpy as np
from scipy import stats

def margin_of_error(data, confidence=0.95):
    """
    Calculate the margin of error for the sample mean.

    Parameters:
    - data: list or numpy array of sample values
    - confidence: confidence level (default is 0.95 for 95%)

    Returns:
    - margin of error (float)
    """
    n = len(data)
    se = stats.sem(data)  # Standard Error of the Mean
    t_critical = stats.t.ppf((1 + confidence) / 2, df=n-1)  # t-critical value
    moe = t_critical * se
    return moe

# Example usage
sample_data = [22, 24, 20, 23, 21, 22, 25, 26, 24, 23]
moe = margin_of_error(sample_data, confidence=0.95)
print(f"Margin of Error at 95% confidence level: {moe:.4f}")
```

```python
#Implement a Bayesian inference method using Bayes' Theorem in Python and explain the process
def bayes_theorem(prior, likelihood, false_alarm_rate):
    """
    Calculate the posterior probability using Bayes' Theorem.

    prior: P(H) - prior probability of hypothesis (disease prevalence)
    likelihood: P(E|H) - probability of evidence given hypothesis (sensitivity)
    false_alarm_rate: P(E|~H) - probability of evidence given not hypothesis (false positive rate)
    """
    # Total probability of evidence P(E)
    evidence = likelihood * prior + false_alarm_rate * (1 - prior)

    # Posterior probability P(H|E)
    posterior = (likelihood * prior) / evidence

    return posterior

# Given data
prior_disease = 0.01          # P(Disease)
sensitivity = 0.99            # P(Positive | Disease)
specificity = 0.95            # P(Negative | No Disease)
false_positive_rate = 1 - specificity  # P(Positive | No Disease)

# Calculate posterior probability of disease given positive test
posterior_prob = bayes_theorem(prior_disease, sensitivity, false_positive_rate)
print(f"Probability of disease given a positive test: {posterior_prob:.4f}")
```

```python
#Perform a goodness-of-fit test using Python to compare the observed data to an expected distribution
from scipy.stats import chisquare
import numpy as np

# Step 1: Observed frequencies (counts)
observed = np.array([50, 30, 20])

# Step 2: Expected frequencies (can be counts or probabilities)
# For example, expected distribution is 0.4, 0.4, 0.2 proportions
expected_proportions = np.array([0.4, 0.4, 0.2])
expected_counts = expected_proportions * observed.sum()  # scale by total observed count

# Step 3: Perform Chi-square goodness-of-fit test
chi2_stat, p_value = chisquare(f_obs=observed, f_exp=expected_counts)

# Step 4: Display results
print(f"Chi-square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Step 5: Interpret
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: Observed data does NOT fit the expected distribution.")
else:
    print("Fail to reject the null hypothesis: Observed data fits the expected distribution.")
```

```python
 #Implement an F-test using Python to compare the variances of two random samples
 import numpy as np
from scipy.stats import f

def f_test(sample1, sample2, alpha=0.05):
    """
    Perform an F-test to compare variances of two samples.

    Parameters:
    - sample1, sample2: arrays or lists of numerical data
    - alpha: significance level (default 0.05)

    Returns:
    - F-statistic, p-value, conclusion string
    """
    # Sample variances
    var1 = np.var(sample1, ddof=1)
    var2 = np.var(sample2, ddof=1)

    # Sample sizes
    n1 = len(sample1)
    n2 = len(sample2)

    # Calculate F-statistic (larger variance / smaller variance)
    if var1 > var2:
        F = var1 / var2
        dfn = n1 - 1  # degrees of freedom numerator
        dfd = n2 - 1  # degrees of freedom denominator
    else:
        F = var2 / var1
        dfn = n2 - 1
        dfd = n1 - 1

    # Calculate two-tailed p-value
    p = 2 * min(f.cdf(F, dfn, dfd), 1 - f.cdf(F, dfn, dfd))

    # Interpretation
    if p < alpha:
        conclusion = "Reject null hypothesis: variances are significantly different."
    else:
        conclusion = "Fail to reject null hypothesis: variances are not significantly different."

    return F, p, conclusion


# Example usage
np.random.seed(0)
sample_a = np.random.normal(loc=10, scale=5, size=30)
sample_b = np.random.normal(loc=10, scale=8, size=35)

F_stat, p_value, result = f_test(sample_a, sample_b)
print(f"F-statistic: {F_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(result)
```

```python
#Write a Python function to check the assumptions (normality, independence, and equal variance) for ANOVA
import numpy as np
from scipy.stats import shapiro, levene
import statsmodels.api as sm
from statsmodels.stats.anova import AnovaRM

def check_anova_assumptions(*groups, check_independence=False):
    """
    Check ANOVA assumptions for multiple groups:
    1. Normality (Shapiro-Wilk test)
    2. Equal variances (Levene's test)
    3. Independence (optional: autocorrelation check)

    Parameters:
    - groups: arrays or lists of numerical data (each group)
    - check_independence: bool, if True will check autocorrelation within each group

    Returns:
    - Dictionary with test results
    """
    results = {}

    # Combine all data for residual normality check
    all_data = np.concatenate(groups)

    # 1. Normality test on each group
    normality_results = {}
    for i, group in enumerate(groups, 1):
        stat, p = shapiro(group)
        normality_results[f'Group_{i}'] = {'statistic': stat, 'p_value': p, 'normal': p > 0.05}
    results['normality'] = normality_results

    # 2. Equal variance test (Levene's test)
    stat, p = levene(*groups)
    results['equal_variance'] = {'statistic': stat, 'p_value': p, 'equal_variance': p > 0.05}

    # 3. Independence (optional) - check autocorrelation in each group
    if check_independence:
        from statsmodels.tsa.stattools import acf
        independence_results = {}
        for i, group in enumerate(groups, 1):
            autocorr = acf(group, fft=False, nlags=1)[1]  # lag-1 autocorrelation
            independence_results[f'Group_{i}'] = {'lag_1_autocorr': autocorr, 'independent': abs(autocorr) < 0.3}
        results['independence'] = independence_results

    return results

# Example usage
group1 = [22, 24, 19, 25, 30]
group2 = [28, 27, 29, 26, 30]
group3 = [35, 37, 34, 40, 38]

assumptions = check_anova_assumptions(group1, group2, group3, check_independence=True)
print(assumptions)
```

```python
#Write a Python program to visualize the F-distribution and discuss its use in hypothesis testing
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f

# Parameters for the F-distribution
dfn = 5  # degrees of freedom numerator
dfd = 10  # degrees of freedom denominator

# Generate x values
x = np.linspace(0, 5, 500)

# Calculate the F-distribution PDF values
pdf = f.pdf(x, dfn, dfd)

# Plot the F-distribution
plt.figure(figsize=(8, 5))
plt.plot(x, pdf, 'b-', lw=2, label=f'F-distribution PDF\n(df1={dfn}, df2={dfd})')
plt.fill_between(x, 0, pdf, alpha=0.2, color='blue')
plt.title('F-Distribution')
plt.xlabel('F value')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Perform a one-way ANOVA test in Python and visualize the results with boxplots to compare group means
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f_oneway

# Sample data: values from three groups
group1 = np.array([22, 24, 19, 25, 30])
group2 = np.array([28, 27, 29, 26, 30])
group3 = np.array([35, 37, 34, 40, 38])

# Perform one-way ANOVA
f_stat, p_value = f_oneway(group1, group2, group3)

print(f"ANOVA F-statistic: {f_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpret results
alpha = 0.05
if p_value < alpha:
    print("Reject null hypothesis: At least one group mean is significantly different.")
else:
    print("Fail to reject null hypothesis: No significant difference between group means.")

# Visualize group distributions with boxplots
plt.boxplot([group1, group2, group3], labels=['Group 1', 'Group 2', 'Group 3'])
plt.title('One-way ANOVA Group Comparison')
plt.ylabel('Values')
plt.show()
```

```python
#Simulate random data from a normal distribution, then perform hypothesis testing to evaluate the means
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Parameters
np.random.seed(42)  # For reproducibility
sample_size = 100
pop_mean = 50
sample_mean = 52
std_dev = 10

# Simulate sample data
sample_data = np.random.normal(loc=sample_mean, scale=std_dev, size=sample_size)

# Perform one-sample t-test
t_stat, p_value = stats.ttest_1samp(sample_data, popmean=pop_mean)

# Results
print(f"Sample Mean: {np.mean(sample_data):.2f}")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The sample mean is significantly different from the population mean.")
else:
    print("Fail to reject the null hypothesis: No significant difference between the sample and population means.")

# Visualize the sample data
plt.figure(figsize=(8, 6))
plt.hist(sample_data, bins=15, edgecolor='black', alpha=0.7, color='skyblue')
plt.axvline(pop_mean, color='red', linestyle='dashed', linewidth=2, label=f'Population Mean = {pop_mean}')
plt.axvline(np.mean(sample_data), color='green', linestyle='dashed', linewidth=2, label=f'Sample Mean = {np.mean(sample_data):.2f}')
plt.title('Histogram of Sample Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Perform a hypothesis test for population variance using a Chi-square distribution and interpret the results
import numpy as np
from scipy.stats import chi2

# Sample data
np.random.seed(42)
sample_data = np.random.normal(loc=50, scale=10, size=30)  # Sample of 30 observations

# Known population variance
population_variance = 100  # Hypothesized population variance

# Calculate sample variance
sample_variance = np.var(sample_data, ddof=1)

# Sample size
n = len(sample_data)

# Degrees of freedom
df = n - 1

# Calculate the chi-square statistic
chi_square_statistic = (df * sample_variance) / population_variance

# Calculate the p-value for a two-tailed test
p_value = 2 * min(chi2.cdf(chi_square_statistic, df), 1 - chi2.cdf(chi_square_statistic, df))

# Significance level
alpha = 0.05

# Interpret the result
if p_value < alpha:
    result = "Reject the null hypothesis: The sample variance is significantly different from the population variance."
else:
    result = "Fail to reject the null hypothesis: No significant difference between the sample and population variances."

# Output the results
print(f"Sample Variance: {sample_variance:.2f}")
print(f"Chi-square Statistic: {chi_square_statistic:.2f}")
print(f"P-value: {p_value:.4f}")
print(result)
```

```python
# Write a Python script to perform a Z-test for comparing proportions between two datasets or groups
import numpy as np
import statsmodels.api as sm

# Sample data: successes and total observations for two groups
# Group 1: 50 successes out of 200 trials
# Group 2: 60 successes out of 250 trials
successes = np.array([50, 60])
trials = np.array([200, 250])

# Perform the two-sample Z-test for proportions
z_stat, p_value = sm.stats.proportions_ztest(successes, trials)

# Display the results
print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: There is a significant difference between the two proportions.")
else:
    print("Fail to reject the null hypothesis: No significant difference between the two proportions.")
```

```python
#Implement an F-test for comparing the variances of two datasets, then interpret and visualize the results
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Sample data: two datasets
np.random.seed(42)
data1 = np.random.normal(loc=50, scale=10, size=100)  # Mean=50, SD=10
data2 = np.random.normal(loc=50, scale=15, size=100)  # Mean=50, SD=15

# Calculate sample variances
var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)

# Degrees of freedom
df1 = len(data1) - 1
df2 = len(data2) - 1

# F-statistic: ratio of variances
F_stat = var1 / var2 if var1 > var2 else var2 / var1

# P-value for two-tailed test
p_value = 2 * min(stats.f.cdf(F_stat, df1, df2), 1 - stats.f.cdf(F_stat, df1, df2))

# Significance level
alpha = 0.05

# Interpretation
if p_value < alpha:
    result = "Reject the null hypothesis: The variances are significantly different."
else:
    result = "Fail to reject the null hypothesis: No significant difference between the variances."

# Output the results
print(f"Sample 1 Variance: {var1:.2f}")
print(f"Sample 2 Variance: {var2:.2f}")
print(f"F-statistic: {F_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(result)

# Visualize the data distributions
plt.figure(figsize=(8, 6))
plt.boxplot([data1, data2], labels=['Sample 1', 'Sample 2'], patch_artist=True)
plt.title('Comparison of Two Samples')
plt.ylabel('Values')
plt.grid(True)
plt.show()
```

```python
#Perform a Chi-square test for goodness of fit with simulated data and analyze the results.
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Simulate observed data (e.g., outcomes of a die roll)
np.random.seed(42)
observed_data = np.random.choice([1, 2, 3, 4, 5, 6], size=600, p=[0.1, 0.2, 0.3, 0.1, 0.2, 0.1])

# Calculate observed frequencies
observed_freq = np.array([np.sum(observed_data == i) for i in range(1, 7)])

# Define expected frequencies (uniform distribution for a fair die)
expected_freq = np.array([100, 100, 100, 100, 100, 100])

# Perform Chi-square goodness-of-fit test
chi2_stat, p_value = stats.chisquare(observed_freq, expected_freq)

# Output the results
print(f"Chi-square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The observed frequencies significantly differ from the expected frequencies.")
else:
    print("Fail to reject the null hypothesis: No significant difference between observed and expected frequencies.")

# Visualize the observed vs expected frequencies
categories = ['1', '2', '3', '4', '5', '6']
x = np.arange(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 6))
rects1 = ax.bar(x - width/2, observed_freq, width, label='Observed', color='skyblue')
rects2 = ax.bar(x + width/2, expected_freq, width, label='Expected', color='salmon')

ax.set_xlabel('Die Face')
ax.set_ylabel('Frequency')
ax.set_title('Observed vs Expected Frequencies')
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend()

plt.tight_layout()
plt.show()
```
