```python
#Write a code to perform basic string compression using the counts of repeated characters
input_string = input("Enter a string: ")

# Initialize variables
compressed = ""
count = 1

# Loop through the string
for i in range(1, len(input_string)):
    if input_string[i] == input_string[i - 1]:
        count += 1
    else:
        compressed += input_string[i - 1] + str(count)
        count = 1

# Add the last character and count
compressed += input_string[-1] + str(count)

# Print the compressed string
print("Compressed string:", compressed)
```

```python
# Implement a code to find the second largest number in a given list of integers
def second_largest(numbers):
    if len(numbers) < 2:
        return None  # Not enough elements

    first = second = float('-inf')
    for num in numbers:
        if num > first:
            second = first
            first = num
        elif first > num > second:
            second = num

    return second if second != float('-inf') else None

# Example usage:
nums = [10, 5, 8, 12, 6, 12]
result = second_largest(nums)
print(result)  # Output: 10
```

```python
#Create a code to count the occurrences of each element in a list and return a dictionary with elements as keys and their counts as values
def count_occurrences(lst):
    counts = {}
    for item in lst:
        counts[item] = counts.get(item, 0) + 1
    return counts

# Example usage:
my_list = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
result = count_occurrences(my_list)
print(result)
# Output: {'apple': 3, 'banana': 2, 'orange': 1}
```

```python
#Write a code to reverse a list in-place without using any built-in reverse functions
def reverse_list_in_place(lst):
    left, right = 0, len(lst) - 1
    while left < right:
        lst[left], lst[right] = lst[right], lst[left]  # Swap elements
        left += 1
        right -= 1

# Example usage:
my_list = [1, 2, 3, 4, 5]
reverse_list_in_place(my_list)
print(my_list)  # Output: [5, 4, 3, 2, 1]
```

```python
# Implement a code to find and remove duplicates from a list while preserving the original order of elements
def remove_duplicates(lst):
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

# Example usage:
my_list = [3, 5, 3, 2, 6, 5, 1, 2]
unique_list = remove_duplicates(my_list)
print(unique_list)  # Output: [3, 5, 2, 6, 1]
```

```python
#Create a code to check if a given list is sorted (either in ascending or descending order) or not
def is_sorted(lst):
    if len(lst) < 2:
        return True  # A list with 0 or 1 element is sorted

    ascending = True
    descending = True

    for i in range(1, len(lst)):
        if lst[i] > lst[i-1]:
            descending = False
        elif lst[i] < lst[i-1]:
            ascending = False

    return ascending or descending

# Example usage:
print(is_sorted([1, 2, 3, 4]))      # True (ascending)
print(is_sorted([4, 3, 2, 1]))      # True (descending)
print(is_sorted([1, 3, 2, 4]))      # False (not sorted)
print(is_sorted([7]))               # True (single element)
print(is_sorted([]))                # True (empty list)
```

```python
# Write a code to merge two sorted lists into a single sorted list
def merge_sorted_lists(list1, list2):
    merged = []
    i, j = 0, 0

    # Traverse both lists and append smaller element to merged
    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            merged.append(list1[i])
            i += 1
        else:
            merged.append(list2[j])
            j += 1

    # Append remaining elements if any
    while i < len(list1):
        merged.append(list1[i])
        i += 1

    while j < len(list2):
        merged.append(list2[j])
        j += 1

    return merged

# Example usage:
l1 = [1, 3, 5, 7]
l2 = [2, 4, 6, 8]
result = merge_sorted_lists(l1, l2)
print(result)  # Output: [1, 2, 3, 4, 5, 6, 7, 8]
```

```python
# Implement a code to find the intersection of two given lists
def intersection(list1, list2):
    set2 = set(list2)
    return [item for item in list1 if item in set2]

# Example usage:
l1 = [1, 2, 3, 4, 5]
l2 = [3, 5, 7, 9]
result = intersection(l1, l2)
print(result)  # Output: [3, 5]
```

```python
#Create a code to find the union of two lists without duplicates
def union_without_duplicates(list1, list2):
    seen = set()
    union_list = []

    for item in list1 + list2:
        if item not in seen:
            seen.add(item)
            union_list.append(item)

    return union_list

# Example usage:
l1 = [1, 2, 3, 4]
l2 = [3, 4, 5, 6]
result = union_without_duplicates(l1, l2)
print(result)  # Output: [1, 2, 3, 4, 5, 6]
```

```python
#Write a code to shuffle a given list randomly without using any built-in shuffle functions
import random

def shuffle_list(lst):
    n = len(lst)
    for i in range(n-1, 0, -1):
        # Pick a random index from 0 to i
        j = random.randint(0, i)
        # Swap elements at i and j
        lst[i], lst[j] = lst[j], lst[i]

# Example usage:
my_list = [1, 2, 3, 4, 5]
shuffle_list(my_list)
print(my_list)
```

```python
#Write a code that takes two tuples as input and returns a new tuple containing elements that are common to both input tuples
def common_elements(tuple1, tuple2):
    set2 = set(tuple2)
    common = [item for item in tuple1 if item in set2]
    return tuple(common)

# Example usage:
t1 = (1, 2, 3, 4, 5)
t2 = (3, 5, 7, 9)
result = common_elements(t1, t2)
print(result)  # Output: (3, 5)
```

```python
# Write a code to access a value in a nested dictionary. The function should take the dictionary and a list of keys as input, and return the corresponding value. If any of the keys do not exist in the dictionary, the
#unction should return None
def merge_dicts(dict1, dict2):
    merged = dict1.copy()  # Start with dict1's items
    for key, value in dict2.items():
        if key in merged:
            merged[key] += value  # Add values if key exists
        else:
            merged[key] = value   # Otherwise, just add the key-value pair
    return merged

# Function to convert user input string to dictionary
def input_to_dict(prompt):
    input_str = input(prompt)
    try:
        # Safely evaluate dictionary from input string
        d = eval(input_str)
        if not isinstance(d, dict):
            raise ValueError
        return d
    except:
        print("Invalid dictionary input. Please enter in format: {'key1': value1, 'key2': value2}")
        exit()

# Get dictionaries from user
dict1 = input_to_dict("Enter the first dictionary (e.g., {'a': 2, 'b': 3}): ")
dict2 = input_to_dict("Enter the second dictionary (e.g., {'b': 4, 'c': 5}): ")
```

```python
#Write a code that takes a dictionary as input and returns a sorted version of it based on the values. You can choose whether to sort in ascending or descending order
def sort_dict_by_values(d, descending=False):
    # Sort dictionary items by value
    sorted_items = sorted(d.items(), key=lambda item: item[1], reverse=descending)
    # Convert back to dictionary
    return dict(sorted_items)

def input_dict():
    while True:
        user_input = input("Enter a dictionary (e.g., {'a': 3, 'b': 1, 'c': 2}): ")
        try:
            d = eval(user_input)
            if not isinstance(d, dict):
                print("Input is not a dictionary. Please try again.")
                continue
            return d
        except:
            print("Invalid dictionary format. Please try again.")

# Get dictionary input from user
dictionary = input_dict()

# Ask user for sorting order
order = input("Sort ascending or descending? (a/d): ").strip().lower()
descending = True if order == 'd' else False

# Get sorted dictionary
sorted_dict = sort_dict_by_values(dictionary, descending)

print("Sorted dictionary:", sorted_dict)
```

```python
#& Write a code that inverts a dictionary, swapping keys and values. Ensure that the inverted dictionary correctly handles cases where multiple keys have the same value by storing the keys as a list in the inverted dictionary
def invert_dict(d):
    inverted = {}
    for key, value in d.items():
        if value in inverted:
            inverted[value].append(key)  # Append to existing list
        else:
            inverted[value] = [key]       # Create new list with this key
    return inverted

# Example usage:
original_dict = {'a': 1, 'b': 2, 'c': 1, 'd': 3}

inverted_dict = invert_dict(original_dict)
print(inverted_dict)
```
