```python
# Explain the following termsg
# The difference in np.array, np.asarray and np.asanyarrayX
# The difference between Deep copy and shallow

- `np.array()`: Creates a new array and copies the data.
- `np.asarray()`: Converts input to array, does NOT copy if input is already array.
- `np.asanyarray()`: Same as asarray but preserves subclass (e.g., matrix subclass).

- Deep Copy: Copies data, changes don't affect original.
- Shallow Copy: Just references original data, changes affect both.
```

```python
#Create a NumPy array with random integers between 1 and 10 of shape (5, 6). After creating the array
  #perform the following operations:

 #a)Extract all even integers from array.

 #b)Extract all odd integers from array

import numpy as np

# Create a 5x6 array with random integers between 1 and 10
array = np.random.randint(1, 11, size=(5, 6))

print("Original array:")
print(array)

# a) Extract all even integers
even_integers = array[array % 2 == 0]

# b) Extract all odd integers
odd_integers = array[array % 2 != 0]

print("\nEven integers:")
print(even_integers)

print("\nOdd integers:")
print(odd_integers)
```

```python
#Create a time-series dataset in a Pandas DataFrame with columns: 'Date', 'Temperature', 'Humidity' and
#Perform the following tasks using Matplotlib:

#a) Plot the 'Temperature' and 'Humidity' on the same plot with different y-axes (left y-axis for 'Temperature' and
#right y-axis for 'Humidity').

#b) Label the x-axis as 'Date'.

#c) Set the title of the plot as 'Temperature and Humidity Over Time'.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create a date range (e.g., 30 days starting from today)
dates = pd.date_range(start=pd.Timestamp.today(), periods=30)

# Generate random temperature and humidity data
temperature = np.random.uniform(low=15, high=35, size=30)  # in degrees Celsius
humidity = np.random.uniform(low=40, high=90, size=30)     # in percentage

# Create the DataFrame
df = pd.DataFrame({
    'Date': dates,
    'Temperature': temperature,
    'Humidity': humidity
})

# Plotting
fig, ax1 = plt.subplots(figsize=(10, 5))

# Plot Temperature on left y-axis
ax1.plot(df['Date'], df['Temperature'], color='red', label='Temperature')
ax1.set_xlabel('Date')
ax1.set_ylabel('Temperature (°C)', color='red')
ax1.tick_params(axis='y', labelcolor='red')

# Create second y-axis for Humidity
ax2 = ax1.twinx()
ax2.plot(df['Date'], df['Humidity'], color='blue', label='Humidity')
ax2.set_ylabel('Humidity (%)', color='blue')
ax2.tick_params(axis='y', labelcolor='blue')

# Set title
plt.title('Temperature and Humidity Over Time')

# Combine legends from both axes
lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
ax1.legend(lines_1 + lines_2, labels_1 + labels_2, loc='upper left')

plt.show()
```

```python
#Create a NumPy array data containing 1000 samples from a normal distribution. Perform the following tasks using Matplotlib:
#a) Plot a histogram of the data with 30 bins.

#b) Overlay a line plot representing the normal distribution's probability density function (PDF).

#c) Label the x-axis as 'Value' and the y-axis as 'Frequency/Probability'.

#d) Set the title of the plot as 'Histogram with PDF Overlay'.

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate 1000 samples from a normal distribution (mean=0, std=1)
data = np.random.randn(1000)

# a) Plot histogram with 30 bins, density=True to normalize
count, bins, ignored = plt.hist(data, bins=30, density=True, alpha=0.6, color='g', label='Histogram')

# b) Overlay normal distribution PDF
# Calculate the PDF values for the bins
mu, std = np.mean(data), np.std(data)
pdf = norm.pdf(bins, mu, std)

plt.plot(bins, pdf, 'r-', lw=2, label='Normal PDF')

# c) Label axes
plt.xlabel('Value')
plt.ylabel('Frequency/Probability')

# d) Set plot title
plt.title('Histogram with PDF Overlay')

# Add legend
plt.legend()

# Show plot
plt.show()
```

```python
#Create a Seaborn scatter plot of two random arrays, color points based on their position relative to the
#origin (quadrants), add a legend, label the axes, and set the title as 'Quadrant-wise Scatter Plot'.

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Generate two random arrays (say 100 points each) between -10 and 10
x = np.random.uniform(-10, 10, 100)
y = np.random.uniform(-10, 10, 100)

# Define a function to determine the quadrant of each point
def quadrant(x_val, y_val):
    if x_val >= 0 and y_val >= 0:
        return 'Quadrant 1'
    elif x_val < 0 and y_val >= 0:
        return 'Quadrant 2'
    elif x_val < 0 and y_val < 0:
        return 'Quadrant 3'
    else:
        return 'Quadrant 4'

# Create a DataFrame
df = pd.DataFrame({'x': x, 'y': y})

# Apply the quadrant function to each row
df['Quadrant'] = df.apply(lambda row: quadrant(row['x'], row['y']), axis=1)

# Create scatter plot with Seaborn, coloring points by Quadrant
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='x', y='y', hue='Quadrant', palette='Set2', s=70)

# Label axes and set title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Quadrant-wise Scatter Plot')

# Show legend and plot
plt.legend(title='Quadrants')
plt.grid(True)
plt.show()
```

```python
#With Bokeh, plot a line chart of a sine wave function, add grid lines, label the axes, and set the title as 'Sine Wave Function
import numpy as np
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

# Enable output in notebook (remove this line if running as script)
output_notebook()

# Generate data for sine wave
x = np.linspace(0, 4 * np.pi, 200)
y = np.sin(x)

# Create a new plot
p = figure(title="Sine Wave Function",
           x_axis_label='x',
           y_axis_label='sin(x)',
           width=700, height=400)

# Add sine wave line
p.line(x, y, line_width=2, color='blue', legend_label='sin(x)')

# Enable grid lines (enabled by default, but you can customize)
p.xgrid.grid_line_color = 'gray'
p.ygrid.grid_line_color = 'gray'
p.xgrid.grid_line_alpha = 0.5
p.ygrid.grid_line_alpha = 0.5

# Show legend
p.legend.location = "top_left"

# Show the plot
show(p)
```

```python
#8 Using Bokeh, generate a bar chart of randomly generated categorical data, color bars based on their
#values, add hover tooltips to display exact values, label the axes, and set the title as 'Random Categorical Bar Chart

import random
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.transform import linear_cmap
from bokeh.palettes import Viridis256
from bokeh.io import output_notebook

# For notebook output; remove if running as script
output_notebook()

# Generate random categorical data
categories = [f'Category {i}' for i in range(1, 11)]
values = [random.randint(10, 100) for _ in categories]

# Prepare the data source
source = ColumnDataSource(data=dict(categories=categories, values=values))

# Create a color mapper based on values
color_mapper = linear_cmap(field_name='values', palette=Viridis256, low=min(values), high=max(values))

# Create figure
p = figure(x_range=categories, height=400, width=700, title="Random Categorical Bar Chart",
           toolbar_location=None, tools="")

# Draw vertical bars with color mapping
bars = p.vbar(x='categories', top='values', width=0.7, source=source, fill_color=color_mapper)

# Add hover tooltips
hover = HoverTool(tooltips=[("Category", "@categories"), ("Value", "@values")], renderers=[bars])
p.add_tools(hover)

# Label axes
p.xaxis.axis_label = "Categories"
p.yaxis.axis_label = "Values"

# Rotate x-axis labels for readability
p.xaxis.major_label_orientation = 1.0

# Show the plot
show(p)
```

```python
# Using Plotly, create a basic line plot of a randomly generated dataset, label the axes, and set the title as 'Simple Line Plot'
import plotly.graph_objects as go
import numpy as np

# Generate random data
x = np.arange(0, 50)
y = np.random.randn(50).cumsum()  # cumulative sum for some trend

# Create figure
fig = go.Figure()

# Add line plot
fig.add_trace(go.Scatter(x=x, y=y, mode='lines', name='Random Data'))

# Update layout with axis labels and title
fig.update_layout(
    title='Simple Line Plot',
    xaxis_title='X-axis',
    yaxis_title='Y-axis'
)

# Show plot
fig.show()
```
